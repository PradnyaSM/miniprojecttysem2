```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re 
def sanitize_filename(filename):
    # Replace invalid characters with an underscore
    return re.sub(r'[<>:"/\\|?*]', '_', filename)

def univariate_analysis(df, output_dir='univariate_analysis'):
    # Create output directory
    os.makedirs(output_dir, exist_ok=True)
    
    # 1. Categorical Variables Analysis
    print("\n=== Categorical Variables Analysis ===")
    
    # a. Age Groups
    if 'Your Age' in df.columns:
        age_freq = df['Your Age'].value_counts(normalize=True).reset_index()
        age_freq.columns = ['Age Group', 'Percentage']
        age_freq['Percentage'] = age_freq['Percentage'] * 100
        print("\nAge Group Distribution:")
        print(age_freq)
        
        plt.figure(figsize=(10, 6))
        sns.barplot(x='Age Group', y='Percentage', data=age_freq)
        plt.title('Age Group Distribution')
        plt.savefig(sanitize_filename(f'{output_dir}/age_distribution.png'))
        plt.close()
    
    # b. Social Media Platforms (Multi-response)
    if 'Which social media platforms do you use most frequently? (Select all that apply)  ' in df.columns:
        platform_counts = df['Which social media platforms do you use most frequently? (Select all that apply)  ']\
            .str.split(', ')\
            .explode()\
            .value_counts()\
            .reset_index()
        platform_counts.columns = ['Platform', 'Count']
        platform_counts['Percentage'] = (platform_counts['Count'] / len(df)) * 100
        
        print("\nSocial Media Platform Popularity:")
        print(platform_counts)
        
        plt.figure(figsize=(12, 8))
        sns.barplot(x='Count', y='Platform', data=platform_counts)
        plt.title('Social Media Platform Popularity')
        plt.tight_layout()
        plt.savefig(sanitize_filename(f'{output_dir}/platform_popularity.png'))
        plt.close()
    
    # 2. Numerical Variables Analysis
    print("\n=== Numerical Variables Analysis ===")
    
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    
    # Create summary statistics table
    stats_table = pd.DataFrame({
        'Variable': numerical_cols,
        'Mean': df[numerical_cols].mean(),
        'Median': df[numerical_cols].median(),
        'Mode': df[numerical_cols].mode().iloc[0],
        'Std Dev': df[numerical_cols].std(),
        'Min': df[numerical_cols].min(),
        'Max': df[numerical_cols].max(),
        'Skewness': df[numerical_cols].skew()
    }).reset_index(drop=True)
    
    print("\nSummary Statistics:")
    print(stats_table)
    
    # Save statistics to CSV
    stats_table.to_csv(sanitize_filename(f'{output_dir}/numerical_summary.csv'), index=False)
    
    # Generate distribution plots
    for col in numerical_cols:
        plt.figure(figsize=(16, 6))
        
        # Histogram
        plt.subplot(1, 2, 1)
        sns.histplot(df[col], kde=True, bins=20)
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        
        # Boxplot
        plt.subplot(1, 2, 2)
        sns.boxplot(x=df[col])
        plt.title(f'Boxplot of {col}')
        plt.xlabel(col)
        
        plt.tight_layout()
        plt.savefig(sanitize_filename(f'{output_dir}/{col}_distribution.png'))
        plt.close()
    
    print(f"\nAnalysis complete. Results saved to {output_dir} directory.")

# Usage example:
# univariate_analysis(cleaned_df)
```

Bivariate Analysis

Cross-tabulations for categorical-categorical relationships

Scatter plots for numerical-numerical relationships

Box plots for categorical-numerical relationships

Correlation matrices

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re
from scipy.stats import chi2_contingency, pearsonr, f_oneway, ttest_ind

def sanitize_filename(filename):
    # Replace invalid characters with an underscore
    filename = re.sub(r'[<>:"/\\|?*\n]', '_', filename)  # Added \n to replace newlines
    # Truncate filename if too long
    filename = filename[:200] if len(filename) > 200 else filename
    return filename

def bivariate_analysis(df, output_dir='bivariate_analysis'):
    os.makedirs(output_dir, exist_ok=True)
    
    # 1. Identify variable types
    numerical_vars = df.select_dtypes(include=np.number).columns.tolist()
    categorical_vars = ['Your Age']  # Add other categorical vars if present
    bool_vars = [col for col in df.columns if df[col].nunique() == 2]
    
    # 2. Categorical-Categorical Analysis
    print("\n=== Categorical-Categorical Analysis ===")
    for i in range(len(categorical_vars)):
        for j in range(i+1, len(categorical_vars)):
            var1 = categorical_vars[i]
            var2 = categorical_vars[j]
            contingency_table = pd.crosstab(df[var1], df[var2])
            chi2, p, _, _ = chi2_contingency(contingency_table)
            print(f"\n{var1} vs {var2}:")
            print(f"Chi2 p-value: {p:.4f}")
            
            plt.figure(figsize=(10, 6))
            sns.heatmap(contingency_table, annot=True, fmt='d', cmap='Blues')
            plt.title(f'{var1} vs {var2}')
            
            # Sanitize and save plot
            filename = sanitize_filename(f'cat_cat_{var1}_vs_{var2}.png')
            plt.savefig(os.path.join(output_dir, filename))
            plt.close()

    # 3. Numerical-Numerical Analysis
    print("\n=== Numerical-Numerical Analysis ===")
    corr_matrix = df[numerical_vars].corr(method='spearman')
    
    # Plot correlation matrix
    plt.figure(figsize=(16, 12))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f",
                mask=np.triu(np.ones_like(corr_matrix, dtype=bool)))
    plt.title('Numerical Variables Correlation Matrix')
    
    # Sanitize and save plot
    filename = sanitize_filename('numerical_correlation_matrix.png')
    plt.savefig(os.path.join(output_dir, filename))
    plt.close()
    
    # Find top correlated pairs
    corr_pairs = corr_matrix.unstack().sort_values(ascending=False)
    corr_pairs = corr_pairs[corr_pairs < 1].drop_duplicates()
    top_pairs = pd.concat([corr_pairs.head(5), corr_pairs.tail(5)])
    
    # Plot top correlations
    for (var1, var2), corr in top_pairs.items():
        plt.figure(figsize=(10, 6))
        sns.regplot(x=var1, y=var2, data=df, scatter_kws={'alpha':0.5})
        plt.title(f'{var1} vs {var2}\nSpearman ρ = {corr:.2f}')
        
        # Sanitize and save plot
        filename = sanitize_filename(f'num_num_{var1}_vs_{var2}.png')
        plt.savefig(os.path.join(output_dir, filename))
        plt.close()

    # 4. Categorical-Numerical Analysis
    print("\n=== Categorical-Numerical Analysis ===")
    for cat_var in categorical_vars:
        for num_var in numerical_vars:
            plt.figure(figsize=(10, 6))
            sns.boxplot(x=cat_var, y=num_var, data=df)
            plt.title(f'{num_var} by {cat_var}')
            plt.xticks(rotation=45)
            
            # Sanitize and save plot
            filename = sanitize_filename(f'cat_num_{cat_var}_vs_{num_var}.png')
            plt.savefig(os.path.join(output_dir, filename))
            plt.close()
            
            # Calculate ANOVA-like statistics
            groups = df.groupby(cat_var)[num_var]
            f_stat, p_value = f_oneway(*[group.dropna() for name, group in groups])
            print(f"\n{cat_var} -> {num_var}:")
            print(f"ANOVA p-value: {p_value:.4f}")

    # 5. Boolean-Numerical Analysis
    print("\n=== Boolean-Numerical Analysis ===")
    for bool_var in bool_vars:
        for num_var in numerical_vars:
            plt.figure(figsize=(10, 6))
            sns.violinplot(x=bool_var, y=num_var, data=df)
            plt.title(f'{num_var} by {bool_var}')
            
            # Sanitize and save plot
            filename = sanitize_filename(f'bool_num_{bool_var}_vs_{num_var}.png')
            plt.savefig(os.path.join(output_dir, filename))
            plt.close()
            
            # T-test between groups
            group0 = df[df[bool_var] == 0][num_var].dropna()
            group1 = df[df[bool_var] == 1][num_var].dropna()
            if len(group0) >= 2 and len(group1) >= 2:
                t_stat, p_value = ttest_ind(group0, group1)
                print(f"\n{bool_var} -> {num_var}:")
                print(f"T-test p-value: {p_value:.4f}")
            else:
                print(f"\n{bool_var} -> {num_var}:")
                print("Not enough valid data for T-test")

    print(f"\nAnalysis complete. Results saved to {output_dir} directory.")

# Usage: bivariate_analysis(cleaned_df)
```

### Phase 3: Hypothesis Testing & Statistical Modeling 
Formulate Hypotheses

Based on initial EDA, create testable hypotheses

Example: "People in age group X are more likely to answer Y to question Z"

```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.stats.power as smp  # Import statsmodels power module
from math import ceil

def sanitize_column_names(df):
    """Sanitize column names by replacing newlines and special characters"""
    df.columns = df.columns.str.replace('\n', '_').str.replace(' ', '_').str.replace('?', '').str.replace('!', '')
    return df

def generate_hypotheses(df, alpha=0.05, min_effect_size=0.3):
    """
    Generate testable hypotheses based on EDA findings
    Returns DataFrame of potential hypotheses with statistical backing
    """
    # Sanitize column names to avoid issues
    df = df.copy()
    df = sanitize_column_names(df)
    
    hypotheses = []
    
    # 1. Age group hypotheses
    if 'Your Age' in df.columns:  # Check if column exists
        age_groups = df['Your Age'].dropna().unique()
        numeric_vars = df.select_dtypes(include=np.number).columns
        
        # Compare age groups across all numerical variables
        for var in numeric_vars:
            if var == 'Your Age':
                continue
                
            # Skip if any group has insufficient data
            groups = []
            valid_groups = []
            for group in age_groups:
                group_data = df[df['Your Age'] == group][var].dropna()
                if len(group_data) >= 2:
                    groups.append(group_data)
                    valid_groups.append(group)
            
            if len(groups) < 2:
                continue
                
            f_stat, p_val = stats.f_oneway(*groups)
            
            if p_val < alpha:
                means = df.groupby('Your Age')[var].mean()
                max_group = means.idxmax()
                min_group = means.idxmin()
                
                hypotheses.append({
                    'type': 'age_group_comparison',
                    'hypothesis': f"People in {max_group} have significantly higher {var} than {min_group}",
                    'p_value': p_val,
                    'effect_size': means.max() - means.min(),
                    'test_used': 'ANOVA'
                })

    # 2. Platform usage hypotheses
    platform_cols = [col for col in df.columns if col.startswith('Uses_')]
    
    for platform in platform_cols:
        for var in df.select_dtypes(include=np.number).columns:
            if var == platform:
                continue
                
            group1 = df[df[platform] == 1][var].dropna()
            group0 = df[df[platform] == 0][var].dropna()
            
            # Skip if any group has insufficient data
            if len(group1) < 2 or len(group0) < 2:
                continue
                
            t_stat, p_val = stats.ttest_ind(group1, group0)
            cohens_d = (group1.mean() - group0.mean()) / np.sqrt(
                (group1.std()**2 + group0.std()**2) / 2
            )
            
            if abs(cohens_d) > min_effect_size and p_val < alpha:
                direction = "higher" if cohens_d > 0 else "lower"
                platform_name = platform.replace('Uses_', '')
                hypotheses.append({
                    'type': 'platform_effect',
                    'hypothesis': f"Users of {platform_name} have {direction} {var}",
                    'p_value': p_val,
                    'effect_size': cohens_d,
                    'test_used': 'T-test'
                })

    # 3. Behavioral hypotheses
    behavior_vars = [
        'Do_you_feel_guilty_about_spending_too_much_time_on_social_media_instead_of_studying',
        'Have_you_ever_sacrificed_sleep_to_stay_on_social_media'
    ]
    
    for behavior in behavior_vars:
        if behavior not in df.columns:
            continue
            
        numeric_vars = df.select_dtypes(include=np.number).columns
        if behavior in numeric_vars:
            numeric_vars = numeric_vars.drop(behavior)
            
        for var in numeric_vars:
            # Calculate correlation
            valid_data = df[[behavior, var]].dropna()
            if len(valid_data) < 2:
                continue
                
            corr, p_val = stats.spearmanr(valid_data[behavior], valid_data[var])
            
            if abs(corr) > min_effect_size and p_val < alpha:
                direction = "positive" if corr > 0 else "negative"
                hypotheses.append({
                    'type': 'behavior_correlation',
                    'hypothesis': f"{behavior} shows {direction} correlation with {var}",
                    'p_value': p_val,
                    'effect_size': corr,
                    'test_used': 'Spearman'
                })

    # Create DataFrame and sort by effect size
    if hypotheses:
        hypotheses_df = pd.DataFrame(hypotheses).sort_values(
            by='effect_size', 
            key=abs, 
            ascending=False
        )
    else:
        hypotheses_df = pd.DataFrame(columns=[
            'type', 'hypothesis', 'p_value', 'effect_size', 'test_used'
        ])
    
    return hypotheses_df

def power_analysis(hypothesis_row, power=0.8):
    """Check if sample size was sufficient for detected effect"""
    if hypothesis_row['test_used'] == 'T-test':
        effect_size = hypothesis_row['effect_size']
        analysis = smp.TTestIndPower()  # Use statsmodels TTestIndPower
        required_n = analysis.solve_power(
            effect_size=effect_size,
            power=power,
            alpha=0.05
        )
        return f"Required sample size per group: {ceil(required_n)}"
    
    elif hypothesis_row['test_used'] == 'ANOVA':
        # Use statsmodels FTestAnovaPower for ANOVA
        k_groups = len(df['Your Age'].unique())  # Number of groups
        effect_size = hypothesis_row['effect_size']
        analysis = smp.FTestAnovaPower()
        required_n = analysis.solve_power(
            effect_size=effect_size,
            alpha=0.05,
            power=power,
            k_groups=k_groups
        )
        return f"Required sample size per group: {ceil(required_n)}"
    
    return "Power analysis not available for this test type"

# Example usage:
# hypotheses_df = generate_hypotheses(df_cleaned)
# hypotheses_df['power_analysis'] = hypotheses_df.apply(power_analysis, axis=1)
# print("Top 10 Testable Hypotheses:")
# print(hypotheses_df.head(10)[['hypothesis', 'effect_size', 'p_value', 'power_analysis']])
```
